```python
categorical_cols = ['Gender','Major','University_Year','Academic_Performance','GitHub_Profile','Leadership_Experience',
                    'LinkedIn_Profile', 'English_Proficiency']
```

```python
df = pd.get_dummies(df, columns=categorical_cols , drop_first=True,dtype=int)
```

```python
y = df['Placement_Status']
X = df.drop(columns=['Student_ID','Company_Tier','Career_Field','Placement_Mode','Starting_Salary_USD','Placement_Status'])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_smote, y_train_smote = smote.fit_resample(
    X_train,
    y_train
)

print("Before SMOTE:")
print(y_train.value_counts())

print("\nAfter SMOTE:")
print(y_train_smote.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

# Logistic Regression
log_smote = LogisticRegression(max_iter=1000)
log_smote.fit(X_train_smote, y_train_smote)
y_pred_log_smote = log_smote.predict(X_test)

# Decision Tree
dt_smote = DecisionTreeClassifier(random_state=42)
dt_smote.fit(X_train_smote, y_train_smote)
y_pred_dt_smote = dt_smote.predict(X_test)

# Random Forest
rf_smote = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)
rf_smote.fit(X_train_smote, y_train_smote)
y_pred_rf_smote = rf_smote.predict(X_test)

# Gradient Boosting
gb_smote = GradientBoostingClassifier(random_state=42)
gb_smote.fit(X_train_smote, y_train_smote)
y_pred_gb_smote = gb_smote.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

models_smote = {
    'Logistic Regression': y_pred_log_smote,
    'Decision Tree': y_pred_dt_smote,
    'Random Forest': y_pred_rf_smote,
    'Gradient Boosting': y_pred_gb_smote
}

smote_results = []

for name, y_pred in models_smote.items():

    smote_results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1 Score': f1_score(y_test, y_pred)
    })

smote_results_df = pd.DataFrame(smote_results)

smote_results_df.sort_values(
    by='F1 Score',
    ascending=False
)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

log_params = {
    'C': [0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs'],
    'max_iter': [1000, 2000]
}

log_grid = GridSearchCV(
    estimator=LogisticRegression(),
    param_grid=log_params,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

log_grid.fit(X_train_smote, y_train_smote)

print("Best Parameters:")
print(log_grid.best_params_)

print("\nBest CV F1 Score:")
print(log_grid.best_score_)
```

```python
best_log_model = log_grid.best_estimator_

y_pred_best_log = best_log_model.predict(X_test)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report
)

print("Tuned Logistic Regression Results")
print("----------------------------------")

print("Accuracy :", accuracy_score(y_test, y_pred_best_log))
print("Precision:", precision_score(y_test, y_pred_best_log))
print("Recall   :", recall_score(y_test, y_pred_best_log))
print("F1 Score :", f1_score(y_test, y_pred_best_log))
print("ROC-AUC  :", roc_auc_score(
    y_test,
    best_log_model.predict_proba(X_test)[:, 1]
))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_best_log))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV

rf_params = {
    'n_estimators': [100, 200, 300],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

rf_grid = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=rf_params,
    cv=5,
    scoring='f1',
    n_jobs=-1,
    verbose=1
)

rf_grid.fit(X_train_smote, y_train_smote)

print("Best Parameters:")
print(rf_grid.best_params_)

print("\nBest CV F1 Score:")
print(rf_grid.best_score_)
```

```python
best_rf_model = rf_grid.best_estimator_

y_pred_best_rf = best_rf_model.predict(X_test)

print("Tuned Random Forest Results")
print("---------------------------")

print("Accuracy :", accuracy_score(y_test, y_pred_best_rf))
print("Precision:", precision_score(y_test, y_pred_best_rf))
print("Recall   :", recall_score(y_test, y_pred_best_rf))
print("F1 Score :", f1_score(y_test, y_pred_best_rf))
print("ROC-AUC  :", roc_auc_score(
    y_test,
    best_rf_model.predict_proba(X_test)[:, 1]
))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_best_rf))
```

```python
final_results = []

final_results.append({
    'Model': 'Tuned Logistic Regression',
    'Accuracy': accuracy_score(y_test, y_pred_best_log),
    'Precision': precision_score(y_test, y_pred_best_log),
    'Recall': recall_score(y_test, y_pred_best_log),
    'F1 Score': f1_score(y_test, y_pred_best_log),
    'ROC-AUC': roc_auc_score(
        y_test,
        best_log_model.predict_proba(X_test)[:, 1]
    )
})

final_results.append({
    'Model': 'Tuned Random Forest',
    'Accuracy': accuracy_score(y_test, y_pred_best_rf),
    'Precision': precision_score(y_test, y_pred_best_rf),
    'Recall': recall_score(y_test, y_pred_best_rf),
    'F1 Score': f1_score(y_test, y_pred_best_rf),
    'ROC-AUC': roc_auc_score(
        y_test,
        best_rf_model.predict_proba(X_test)[:, 1]
    )
})

final_results_df = pd.DataFrame(final_results)

final_results_df.sort_values(
    by='F1 Score',
    ascending=False
)
```
